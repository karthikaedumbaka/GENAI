```python
import os
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel ,Field
from langchain.schema.runnable import RunnableParallel,RunnableBranch,RunnableLambda,RunnableSequence,RunnablePassthrough
from typing import Literal
load_dotenv()
```

```python
prompt = PromptTemplate(
    template='make a best Joke about{topic}',
    input_variables=['topic']
)
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    google_api_key=os.environ["GEMINI_API_KEY"]
)
```

```python
parser = StrOutputParser()
```

```python
def word_count(text):
    return len(text.split())
```

```python
joke_chain= RunnableSequence(prompt,model,parser)
```

```python
parallel_chain = RunnableParallel({
    "joke":RunnablePassthrough(),
    'word_count':RunnableLambda(word_count)
})
```

```python
final_chain = RunnableSequence(joke_chain , parallel_chain)
```

```python
final_chain.invoke({"topic": "YouTube"})
```

```python
final_chain.get_graph().print_ascii()
```
