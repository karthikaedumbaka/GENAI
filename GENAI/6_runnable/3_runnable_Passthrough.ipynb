```python
import os
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel ,Field
from langchain.schema.runnable import RunnableParallel,RunnableBranch,RunnableLambda,RunnableSequence,RunnablePassthrough
from typing import Literal
load_dotenv()
```

```python
passthrough= RunnablePassthrough()
```

```python
passthrough.invoke(2)
```

```python
passthrough.invoke("klay")
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    google_api_key=os.environ["GEMINI_API_KEY"]
)
```

```python
prompt1 = PromptTemplate(
    template="Write a joke about {topic}",
    input_variables=['topic']
)


prompt2 = PromptTemplate(
    template="Expain a joke about {topic} in 3 points",
    input_variables=['topic']
)
```

```python
parser= StrOutputParser()
```

```python
joke_gen_chain = RunnableSequence(prompt1,model,parser)
```

```python
parallel_chain = RunnableParallel({
    "joke":RunnablePassthrough(),
    "explanation": RunnableSequence(prompt2,model,parser)
})
```

```python
final_chain = RunnableSequence(joke_gen_chain,parallel_chain)
```

```python
final_chain.invoke("Food")
```

```python
final_chain.get_graph().print_ascii()
```
