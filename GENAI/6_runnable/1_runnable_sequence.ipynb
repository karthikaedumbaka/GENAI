```python
import os
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel ,Field
from langchain.schema.runnable import RunnableParallel,RunnableBranch,RunnableLambda,RunnableSequence
from typing import Literal
load_dotenv()
```

```python
# Model
model = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    google_api_key=os.environ["GEMINI_API_KEY"]
)
```

```python
prompt = PromptTemplate(
    template="Write a joke about {topic}",
    input_variables=['topic']
)
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | model |parser
```

```python
chain.invoke({"topic","GenAI"})
```

```python
prompt2 = PromptTemplate(
    template="Expain a joke about {topic} in 3 points",
    input_variables=['topic']
)
```

```python
chain = prompt | model |parser | prompt2 | model |parser
```

```python
chain.invoke({"topic","GenAI"})
```
