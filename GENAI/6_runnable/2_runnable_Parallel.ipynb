```python
import os
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel ,Field
from langchain.schema.runnable import RunnableParallel,RunnableBranch,RunnableLambda,RunnableSequence
from typing import Literal
load_dotenv()
```

```python
# Model
model1 = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    google_api_key=os.environ["GEMINI_API_KEY"]
)
```

```python
# Model
model2 = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    google_api_key=os.environ["GEMINI_API_KEY"]
)
```

```python
prompt1 = PromptTemplate(
    template="Generate a Linkedin poat about{topic}",
    input_variables=['topic']
)
```

```python
prompt2 = PromptTemplate(
    template="Generate a twwet poat about{topic}",
    input_variables=['topic']
)
```

```python
parser = StrOutputParser()
```

```python
parallel_chain =RunnableParallel({
    "tweet":RunnableSequence(prompt2 , model2,parser),
    "linkedin":RunnableSequence(prompt1,model1,parser)
})
```

```python
result = parallel_chain.invoke({"topic":"F1"})
```
